# 一、series学习

## 1.series的创建

```python
from os import name

import numpy as np
# series的创建
import pandas as pd
s = pd.Series([1,2,3,4,5])
print(s)
```

```python
# 自定义索引
s = pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
print(s)
# 定义name
s = pd.Series([1,2,3,4,5],index=['a','b','c','d','e'],name = '月份')
print(s)
```

```python
# 通过字典形式创建
s = pd.Series({'a':1,'b':2,'c':3,'d':4} ,name = '月份' )
print(s)
s1 = pd.Series(s,index=['a','b'])
print(s1)
```

## 2.Series的属性

```python
'''
index:Series的索引对象
values:Series的值
dtype或dtypes:Series的元素类型
shape:Series的形状
ndim:Series的维度
size:Series的元素个数
name:Series的名称
Loc[] 显式索引，按标签索引或切片
iloc[] 隐式索引，按位置索引或切片
at[] 使用标签访问单个元素
iat[] 使用位置访问单个元素
'''
print(s)
print(s.index)
print(s.values)
print(s.shape)
print(s.ndim)
print(s.size)
print(s.name)
print(s.loc['a']) #显式索引，支持切片
print(s.iloc[1]) #隐式索引，支持切片
print(s.at['a']) # 不支持切片
print(s.iat[0])
```

```python
# 访问数据
print(s['a'])
print(s[s<3])
s = pd.Series([1,2,3,4,5,6,7,8,9],index=['a','b','c','d','e','f','g','h','i'],name='月份')
print(s)
print(s.head()) #打印前五行
print(s.head(2)) #打印前2行
print(s.tail()) # 打印末尾五行
print(s.tail(2)) # 打印末尾2行
```

```python
# 常见函数
s = pd.Series([1,np.nan,3,None,5,6,7,8,9],index=['a','b','c','d','e','f','g','h','i'],name='月份')
print(s)
```

## 2.案例练习

```python
text_data = [
    "大模型RAG可以实现私有知识库问答",
    "向量数据库负责文本相似度检索",
    "LangChain提供文档分割与检索链工具",
    "Ollama支持本地离线部署开源大模型",
    "FastAPI可以将RAG封装成接口服务"
]
text_series = pd .Series(text_data)
# 构建Series
text_series = pd.Series(text_data, name="知识库文本")
print("=== 案例1：基础文本Series ===")
print(text_series)
print("-" * 80)

# ======================
# 案例2：自定义索引Series（给每条文本分配文档ID）
# ======================
doc_ids = ["doc001", "doc002", "doc003", "doc004", "doc005"]
index_series = pd.Series(text_data, index=doc_ids, name="带文档ID的文本")
print("=== 案例2：自定义索引Series ===")
print(index_series)
print("通过文档ID doc003取值：", index_series["doc003"])
print("-" * 80)

# ======================
# 案例3：Series基础取值、切片（筛选知识库片段）
# ======================
print("=== 案例3：切片与取值操作 ===")
# 取前2条文本
print("前两条文本：")
print(index_series[:2])
# 条件筛选：包含RAG关键词的文本
rag_text = index_series[index_series.str.contains("RAG")]
print("\n包含RAG关键词的文本：")
print(rag_text)
print("-" * 80)

# ======================
# 案例4：Series文本清洗（RAG核心需求：文本预处理）
# ======================
dirty_text = [
    "  向量数据库，用于文本检索！！  ",
    "LANGCHAIN 实现检索问答链##",
    "  Ollama本地大模型部署  ",
    "FAISS 轻量级向量库，适合本地项目！"
]
dirty_series = pd.Series(dirty_text, name="待清洗文本")
print("=== 案例4：文本清洗实战 ===")
print("清洗前：")
print(dirty_series)

# 清洗步骤：去空格、转小写、去除特殊符号
clean_series = dirty_series.str.strip()  # 去除首尾空格
clean_series = clean_series.str.lower()  # 全部转为小写
clean_series = clean_series.str.replace(r"[#！]", "", regex=True)  # 去除符号
print("\n清洗后：")
print(clean_series)
```

## 4.案例练习

```python
import pandas as pd

# 模拟私有知识库问答数据
data = {
    "doc_id": ["doc001", "doc002", "doc003", "doc004", "doc005", "doc006"],
    "category": ["RAG基础", "向量库", "大模型部署", "RAG基础", "向量库", "大模型部署"],
    "question": [
        "什么是RAG检索增强生成？",
        "FAISS和Chroma有什么区别？",
        "Ollama怎么本地运行Llama3？",
        "RAG可以解决大模型幻觉吗？",
        "向量数据库的相似度检索原理？",
        "如何限制Ollama占用显存？"
    ],
    "answer": [
        "RAG通过检索外部知识库再交给大模型生成答案，减少幻觉。",
        "FAISS轻量适合本地；Chroma开箱即用，自带持久化存储。",
        "安装Ollama后终端执行 ollama run llama3 即可本地调用。",
        "可以，RAG让模型基于真实文档回答，凭空编造内容大幅减少。",
        "文本转为向量，通过余弦距离计算向量相似度，匹配相似文档。",
        "启动模型时添加参数 num_gpu 限制显卡显存分配。"
    ],
    "char_length": [36, 42, 39, 35, 41, 38]
}

# 创建DataFrame
df = pd.DataFrame(data)
print("完整知识库问答表格：")
df
```

```python
# 1. 查看所有RAG基础分类的数据
rag_base = df[df["category"] == "RAG基础"]
print("==== RAG基础类问答 ====")
display(rag_base)

# 2. 多条件筛选：向量库分类 + 字符长度大于40
vector_long = df[(df["category"] == "向量库") & (df["char_length"] > 40)]
print("\n==== 向量库长文本问答 ====")
display(vector_long)

# 3. 只提取问题、答案两列
qa_df = df[["question", "answer"]]
print("\n==== 仅问答字段 ====")
display(qa_df)
```

```python
# 制造带缺失值的数据（模拟爬取/导入空白文档）
dirty_data = {
    "doc_id": ["doc007", "doc008", "doc009"],
    "category": ["向量库", None, "RAG基础"],
    "question": ["Chroma如何持久化数据？", "本地RAG怎么搭建？", ""],
    "answer": ["使用persist_directory参数保存向量文件", "分文档加载、向量化、检索三步", None]
}
dirty_df = pd.DataFrame(dirty_data)
print("清洗前（存在空值、空白文本）：")
display(dirty_df)

# 清洗步骤
# 1. 空白question替换为NaN
dirty_df["question"] = dirty_df["question"].replace("", pd.NA)
# 2. 删除answer为空的无效数据
clean_df = dirty_df.dropna(subset=["answer"])
# 3. 分类缺失值填充为"未分类"
clean_df["category"] = clean_df["category"].fillna("未分类")

print("\n清洗完成后的有效知识库数据：")
display(clean_df)
```

```python
# 按文档分类分组统计
group_result = df.groupby("category").agg(
    文档数量=("doc_id", "count"),
    平均文本长度=("char_length", "mean")
).round(2)

print("各知识库分类统计汇总：")
display(group_result)
```

```python
# 合并原始干净数据和清洗数据
full_kb = pd.concat([df, clean_df], ignore_index=True)
# 导出csv文件，保存到本地，后续读取做向量入库
full_kb.to_csv("rag_knowledge_base.csv", index=False, encoding="utf-8-sig")
print("知识库已导出为 rag_knowledge_base.csv")

# 读取csv验证
load_df = pd.read_csv("rag_knowledge_base.csv", encoding="utf-8-sig")
print("\n读取导出后的文件：")
display(load_df.head())
```
