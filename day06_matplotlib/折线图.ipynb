```python


import matplotlib.pyplot as plt
from matplotlib import rcParams #控制字体
rcParams['font.family'] = ['SimHei']
# 创建图表，设置大小
plt.figure(figsize=(10,5))
# 要绘制的图表数据
month = ['一月','二月','三月','四月']
sales = [100,150,80,130]
# 绘制折线图
plt.plot(month,sales,label='产品A',color='orange',linewidth=2,linestyle='--',marker='o')
plt.title('2025年销售趋势',color='red',fontsize=50)
# 添加坐标轴的标签
plt.xlabel('月份',fontsize=20)
plt.ylabel('销售额(万元)',fontsize=20)
# 添加图例
plt.legend(loc='upper left',fontsize=10)
# 添加网格线
plt.grid(True,alpha=0.1,color='green',linestyle='--')
# plt.grid(axis='x')
# 设置刻度字体大小
plt.xticks(rotation=0,fontsize=20)
plt.yticks(rotation=10,fontsize=20)
# 设置y轴范围
plt.ylim(0,160)
# 在每个数据点上显示数值
for x,y in zip(month,sales):
    print(x,y)
    plt.text(x,y+5,str(y),ha='center',va='center')
# 显示图表
plt.show()
```
