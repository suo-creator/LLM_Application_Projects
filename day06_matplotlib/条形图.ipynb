```python
from cProfile import label
from calendar import month

import matplotlib.pyplot as plt
from matplotlib import rcParams
from numpy.ma.core import left_shift
# 控制字体
rcParams['font.family'] = ['SimHei']
# 创建图表，设置大小
plt.figure(figsize=(10, 5))
# 要绘制的图表数据
subject = ['语文', '数学', '英语', '科学']
score = [85,92,78,88]

# 绘制条形图
plt.barh(subject, score,label='小明',color='yellow')
# 添加标题
plt.title('成绩分布图', color='red', fontsize=50)
# 添加坐标轴的标签
plt.xlabel('成绩', fontsize=20)
plt.ylabel('分数', fontsize=20)
# 添加图例
plt.legend(loc='upper right', fontsize=8)
# 添加网格线
plt.grid(axis='x', alpha=0.5, color='green', linestyle='--')
# plt.grid(axis='x')
# 设置刻度字体大小
plt.xticks(rotation=0, fontsize=20)
plt.yticks(rotation=10, fontsize=20)
# 设置x轴范围
plt.xlim(0, 160)
# 在每个数据点上显示数值
for x, y in zip(score, subject):
    print(x, y)
    plt.text(x+5, y, str(x), ha='center', va='center')
# 自动优化排版
plt.tight_layout()
# 显示图表
plt.show()
```

```python
import pandas as pd
from matplotlib import pyplot as plt
from matplotlib import rcParams
plt.rcParams['font.family'] = ['SimHei']
df = pd.read_csv('rag_knowledge_base.csv')
df
```

```python
df['answer_length'] = df['answer'].str.len()
df
```

```python
category_count = df['category'].value_counts()
plt.figure(figsize=(10, 5))
plt.bar(category_count.index,category_count.values)
plt.title('RAG知识库——各分类文档数量')
plt.xlabel('文档分类',fontsize=15)
plt.ylabel('文档数量',fontsize=15)
plt.xticks(rotation=0, fontsize=10)
plt.yticks(rotation=0, fontsize=10)
plt.ylim(0,4)
for x, y in zip(category_count.index,category_count.values):
    print(x, y)
    plt.text(x, y+0.1, str(y), ha='center', va='center')
plt.tight_layout()
plt.show()
```
